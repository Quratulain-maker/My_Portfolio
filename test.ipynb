```python
import os
from pathlib import Path

def print_directory_tree(path, prefix=""):
    """
    Recursively prints a visual tree of the directory structure.
    """
    path = Path(path)

    # Base case: check if the initial path exists
    if not path.exists():
        print(f"Error: The path '{path}' does not exist.")
        return

    # Print the root directory name
    if prefix == "":
        # Handle cases where the path is a drive root (e.g., 'C:\' or '/')
        root_name = path.name if path.name else str(path)
        print(f"📁 {root_name}/")

    try:
        # Gather all items in the current directory
        items = list(path.iterdir())
    except PermissionError:
        # Handle folders that require admin/root privileges
        print(f"{prefix}└── 🚫 [Access Denied]")
        return

    # Sort items: Directories first, then files, both alphabetically
    items.sort(key=lambda x: (not x.is_dir(), x.name.lower()))

    # Create the branch pointers
    pointers = ["├── "] * (len(items) - 1) + ["└── "] if items else []

    for pointer, item in zip(pointers, items):
        if item.is_dir():
            # Print the directory and recurse into it
            print(f"{prefix}{pointer}📁 {item.name}/")

            # Determine the prefix for the next level based on the current pointer
            extension = "│   " if pointer == "├── " else "    "
            print_directory_tree(item, prefix=prefix + extension)
        else:
            # Print the file with its extension
            print(f"{prefix}{pointer}📄 {item.name}")

if __name__ == "__main__":
    print("--- Directory Tree Generator ---")
    folder_input = input("Enter the full path of the folder: ")

    # Strip any accidental quotes from dragging and dropping folders into the terminal
    folder_input = folder_input.strip('\"\'')

    print("\nGenerating tree...\n")
    print_directory_tree(folder_input)
```
